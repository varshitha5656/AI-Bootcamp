```python
import requests
API_KEY = "3206b2dd1fc76c780550a95cbf0b199c"
movie = input("Enter movie name: ")

url="https://api.themoviedb.org/3/search/movie"
parameters = {
    "query" : movie,
    "api_key" : API_KEY,
}
response = requests.get(url, params=parameters)
print(response.status_code)
data = response.json()
print("Movie: ", data["results"][0]["title"])
print("Release data: ", data["results"][0]["release_date"])
print("Rating: ", data["results"][0]["vote_average"])
print(data["results"][0]["overview"])
```
