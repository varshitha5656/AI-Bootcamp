```python
word = input("Enter a string: ")
counts={}
for char in word:
  if char != " ":
    if char in counts:
      counts[char] = counts[char] + 1
    else:
      counts[char] = 1
for char in word:
  if char != " ":
    if counts[char] == 1:
      print("First non-repeating character:", char)
      break
else:
  print("No non-repeating character found")
```
